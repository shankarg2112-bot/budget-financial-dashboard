```python
#Department-level variance -This should show Marketing as the worst overspender.
df = pd.read_csv('Budget vs Actual.csv')

dept_summary = df.groupby('Department').agg(
    Total_Budget=('Budget Amount','sum'),
    Total_Actual=('Actual Amount','sum')
).reset_index()

dept_summary['Variance'] = dept_summary['Total_Actual'] - dept_summary['Total_Budget']
dept_summary['Variance %'] = (dept_summary['Variance'] / dept_summary['Total_Budget'] * 100).round(2)
dept_summary = dept_summary.sort_values('Variance %', ascending=False)
dept_summary
```

```python
#Regional level Variance
region_summary = df.groupby('Region').agg(
    Total_Budget=('Budget Amount','sum'),
    Total_Actual=('Actual Amount','sum')
).reset_index()

region_summary['Variance'] = region_summary['Total_Actual'] - region_summary['Total_Budget']
region_summary['Variance %'] = (region_summary['Variance'] / region_summary['Total_Budget'] * 100).round(2)
region_summary = region_summary.sort_values('Variance %', ascending=False)
region_summary
```

```python
#Department and Region wise variance
pivot = df.groupby(['Department','Region']).apply(
    lambda g: round((g['Actual Amount'].sum() - g['Budget Amount'].sum()) / g['Budget Amount'].sum() * 100, 2)
).unstack()
pivot
```
